```python
output_cols = [
    'gene_id',  'logFC', 'AveExpr', 'pvalue', 'padj', 'Assay',
    'contrast1', 
]
```

```python
for contrast, fname in contrast_map.items():
    fpath = os.path.join(out_dpath, f'{fname}.csv')
    df.loc[df['contrast1'].isin([contrast]), output_cols].to_csv(fpath, index=False)
```

```python
def _contrast_to_legend(contrast):
    contrast_fname = contrast_map[contrast]
    parts = contrast_fname.split('__')

    parts = [p.replace('_', ' ') for p in parts]
    organism , assay, treatment, day, _, ref_day = parts
    legend = f'DE genes of {organism} when {day} is compared to {ref_day}. Both samples are using {treatment}. {assay} data. '
    return legend

_contrast_to_legend('C1vsP1_rna')
```

```python
for contrast in contrast_map:
    print(f"{contrast}: {_contrast_to_legend(contrast)}\n")
```

```python
legend_fpath = os.path.join(out_dpath, 'MED4_legend.txt')
with open(legend_fpath, 'w') as f:
    for contrast in contrast_map:
        f.write(f"{contrast_map[contrast]}.csv: {_contrast_to_legend(contrast)}\n\n")
```
